## Exercise 1 — Comparative Analysis of Generative AI vs. Traditional AI

### 1. Automated Medical Diagnosis
**Best choice:** Traditional AI (classification models)

**Justification:**
- Medical diagnosis requires **high accuracy, reliability, and explainability**.
- Traditional AI models (CNNs, supervised classifiers) perform exceptionally well on **image classification**.
- Generative AI can augment training data, but using generative models for diagnosis introduces **bias, hallucinations, and lack of interpretability**, which are unsafe in healthcare.
- Regulatory compliance (FDA, CE) demands **transparent decision-making**, which generative models lack.

---

### 2. Legal Document Generation
**Best choice:** Generative AI (LLMs)

**Justification:**
- Legal contracts require **complex language understanding and generation**, which rule-based systems cannot handle.
- Generative AI can draft coherent documents, adapt writing style, and ensure logical structure.

**Risks:**
- Hallucinated clauses
- Incorrect legal terminology
- Bias or unsafe recommendations
- Liability issues

**Conclusion:** Generative AI is suitable, but requires **human review**, legal validation tools, and safety alignment.

---

### 3. AI-Generated Scientific Research
**Best choice:** Generative AI (with constraints)

**Justification:**
- Summarizing literature → **LLMs perform extremely well**.
- Suggesting hypotheses → generative models can propose ideas based on patterns.

**Concerns:**
- Risk of generating **false citations**, hallucinated facts.
- Lack of understanding of experimental feasibility.
- Overconfidence in wrong outputs.

Generative AI can help, but **must be supervised by domain experts**.

---

### 4. Financial Market Predictions
**Best choice:** Traditional AI models

**Justification:**
- Finance prediction relies on **time-series forecasting**, statistical patterns, risk modeling.
- Generative AI is **not reliable** due to:
  - hallucinations  
  - inability to respect strict numerical constraints  
  - dangerous unpredictability

Regression, ARIMA, LSTMs, and other deterministic models are more appropriate.

---

### 5. Autonomous Vehicle Decision-Making
**Best choice:** Traditional AI (deterministic models, rule-based logic)

**Justification:**
- AVs need **real-time, predictable, explainable** decisions.
- Generative models are **too unpredictable**, may hallucinate impossible scenarios.
- Critical safety tasks require:
  - sensor fusion  
  - object detection  
  - trajectory planning  

Generative AI can help simulate training environments but **should not drive decision logic**.


## Exercise 2 — Ethical and Security Risks of Generative AI

---

### 1. Deepfake Political Manipulation
**Risks:**
- Large-scale misinformation during elections
- Loss of trust in media
- Inciting political violence

**Mitigations:**
- Mandatory watermarking of AI-generated media
- Election-period content authenticity verification systems

---

### 2. Synthetic Identity Fraud
**Risks:**
- AI-generated faces bypassing biometric systems
- Massive financial theft
- Undetectable criminal activity

**Mitigations:**
- Multi-factor authentication combining biometrics + device signals
- Robust deepfake detection models trained on synthetic identities

---

### 3. Generative AI in Cyber Warfare
**Risks:**
- Fake intelligence reports influencing diplomacy
- Manipulation of global conflict decisions
- Undermining national security

**Mitigations:**
- Provenance metadata for classified documents
- Secure channels with cryptographically verifiable origin

---

### 4. AI-Generated Malware
**Risks:**
- Infinite variants of malware
- Automated exploitation discovery
- Evasion of signature-based detection

**Mitigations:**
- Behavioral anomaly detection systems
- AI-powered malware classifiers with constant retraining

---

### 5. Copyright and IP Theft
**Risks:**
- Models copying protected content
- Legal liability for unauthorized use
- Creators losing income

**Mitigations:**
- Training on licensed datasets only
- Copyright-respecting filters for generated content


## Exercise 3 — Optimization and Fine-Tuning of Generative AI

---

### 1. Prompt Engineering (rewrite)

- Original: “Generate an image of a futuristic city.”
  → Improved: “Generate a highly detailed 4K illustration of a futuristic cyberpunk city at night, with neon lights, flying vehicles, and dense skyscrapers.”

- Original: “Write a poem about the future.”
  → Improved: “Write a 12-line rhyming poem about the future of humanity, focusing on hope, technology, and exploration.”

- Original: “Create a song in the style of classical music.”
  → Improved: “Compose a 1-minute classical orchestral piece inspired by Mozart, with strings leading the melody and a soft piano accompaniment.”

---

### 2. Fixing Bias in Training Data
Solutions:
- Add diverse, politically balanced sources.
- Debias text using reweighting or adversarial debiasing.
- Human auditing for bias themes.
- Remove extreme partisan content.

---

### 3. Fine-Tuning for Domain-Specific Medical Research
Steps:
1. Collect peer-reviewed medical datasets.
2. Clean and normalize terminology.
3. Use transfer learning on an LLM already trained on scientific text.
4. Apply strict factual accuracy metrics (FactualityScore, SciBERTScore).
5. Human expert evaluation before deployment.

---

### 4. Quantitative Metrics for Generative AI
- **BLEU** → text similarity metric
- **ROUGE** → recall-based summary comparison
- **Perplexity** → model confidence
- **FID (Fréchet Inception Distance)** → image realism
- **CLIPScore** → image-text alignment

---

### 5. Controlling Creativity & Coherence
- **Temperature scaling**: low = coherent, high = creative
- **Reinforcement learning**: align with desired behaviors
- **Attention mechanisms**: maintain context and narrative coherence

## Exercise 5 — Latent Space Exploration in VAEs

---

### 1. Visualizing Latent Space Distributions
Evaluate by:
- t-SNE or UMAP projections
- Clustering (k-means) to see if digits form distinct clusters
- Latent space heat maps

---

### 2. Interpolating Between Samples
Steps:
- Encode image A → z₁
- Encode image B → z₂
- Linearly interpolate: z(t) = (1-t)z₁ + t z₂
- Decode each z(t)

This is possible because VAEs enforce a **continuous, smooth latent space**.  
GANs lack this structure.

---

### 3. Controlling KL Divergence
- Increase KL → smoother latent space, less variety
- Decrease KL → more diversity but more noise

Real-world example: medical imaging (low variability required)

---

### 4. Disentangling Latent Representations
Modify a single latent dimension corresponding to:
- hair color
- expression
- age

Keep others constant → controlled attribute editing.

---

### 5. PCA vs. VAEs
| Aspect | PCA | VAE |
|--------|------|------|
| Linearity | Linear | Non-linear |
| Interpretability | High | Medium |
| Reconstruction | Poor for complex data | Very good |
| Generative Ability | None | Strong generative model |
