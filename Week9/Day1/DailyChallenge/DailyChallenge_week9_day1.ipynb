```python
summary = """
Machine Learning (ML) is a branch of AI that allows computers to learn from data
and make predictions or decisions without explicit programming.
For data analysts, ML is crucial because it enables forecasting, classification,
and pattern detection, going beyond descriptive statistics.
"""
print(summary)
```

```python
applications = {
    "Finance": "Fraud detection in credit card transactions",
    "Healthcare": "Medical image analysis and disease prediction",
    "Retail/E-commerce": "Recommendation systems like Amazon/Netflix",
}

print("🌍 Applications of ML:")
for k, v in applications.items():
    print(f"- {k}: {v}")
```

```python
types_ml = {
    "Supervised Learning": "Learn from labeled data. Example: Predict loan default.",
    "Unsupervised Learning": "Learn from unlabeled data, find hidden structures. Example: Customer segmentation.",
    "Reinforcement Learning": "Agent learns by interacting with environment via rewards/penalties. Example: Robot navigating a maze.",
}

print("🤖 Types of Machine Learning:")
for k, v in types_ml.items():
    print(f"- {k}: {v}")
```

```python
process_ml = """
1. Feature Selection:
   Identify the most relevant variables (e.g., credit score, income for loan default).
2. Model Selection:
   Choose the right algorithm (Logistic Regression, Decision Tree, Neural Network...).
3. Model Evaluation:
   Test the model with metrics like Accuracy, Precision, Recall, F1-score, AUC.
"""
print("🔄 Process of a ML model:")
print(process_ml)
```
