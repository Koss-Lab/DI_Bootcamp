```python
# Exercise 1 - Problem & Data Collection

problem_statement = """
We want to predict whether a loan applicant will default or not.
This is a binary classification problem (Default = 1, No Default = 0).
"""

data_needs = [
    "Applicant personal details (age, gender, marital status, dependents, education, employment)",
    "Financial info (income, coapplicant income, loan amount, loan term, property area)",
    "Credit history (credit score, past defaults, repayment history)"
]

data_sources = [
    "Internal bank records",
    "Credit bureaus (Experian, Equifax, etc.)",
    "Public financial datasets"
]

print(problem_statement)
print("Data types needed:", data_needs)
print("Possible data sources:", data_sources)
```

```python
# Exercise 2 - Feature Selection

import pandas as pd
from google.colab import files

# Upload CSV from your computer
uploaded = files.upload()

# Load the first uploaded file
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

# Select features
features_selected = [
    "ApplicantIncome", "CoapplicantIncome", "LoanAmount",
    "Loan_Amount_Term", "Credit_History", "Dependents",
    "Education", "Property_Area", "Self_Employed"
]

print("Selected features for loan default prediction:")
print(features_selected)

print("\nJustification: Income and credit history are strong predictors of repayment ability.")
print("Property_Area and Employment can also influence default risk.")
print("Loan_ID dropped as it's just an identifier.")
```

```python
# Exercise 3 - Model Training & Evaluation

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

# Drop ID + separate target
df = df.drop("Loan_ID", axis=1)
y = (df["Loan_Status"] == "Y").astype(int)
X = df.drop("Loan_Status", axis=1)

# Encode categorical
for col in X.select_dtypes(include="object").columns:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))

# Handle missing values
imputer = SimpleImputer(strategy="median")
X[X.columns] = imputer.fit_transform(X)

# Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# Models
log_reg = LogisticRegression(max_iter=1000)
rf = RandomForestClassifier(random_state=42)

# Train & evaluate Logistic Regression
log_reg.fit(X_train, y_train)
print("Logistic Regression Report:\n", classification_report(y_test, log_reg.predict(X_test)))

# Train & evaluate Random Forest
rf.fit(X_train, y_train)
print("Random Forest Report:\n", classification_report(y_test, rf.predict(X_test)))
```
