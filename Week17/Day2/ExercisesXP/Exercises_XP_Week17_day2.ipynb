```python
# TODO: capture definitions and key characteristics
open_source_levels = {
    "Fully Open": {
        "definition": (
            "A model release where the training code (or full source), model architecture, and pretrained weights "
            "are available under an OSI-style open-source license (or similarly permissive terms), enabling full inspection "
            "and modification."
        ),
        "what_is_open": [
            "Training/inference code + model architecture details",
            "Pretrained weights + (often) configs/tokenizer and sometimes training recipes"
        ],
        "what_you_can_do": [
            "Inspect internals, modify code/architecture, reproduce runs, fine-tune and (if data/compute) retrain",
            "Deploy locally or in production with maximum control (subject to license obligations)"
        ],
        "what_you_cannot_do": [
            "Guarantee reproducibility if training data/compute details are missing",
            "Ignore privacy/safety/compliance: you still must follow data protection + sector regulations"
        ],
    },
    "Weights Released": {
        "definition": (
            "A model release where pretrained weights are available (often with inference code/configs), "
            "but full training pipeline and/or data details may be incomplete or not released."
        ),
        "what_is_open": [
            "Pretrained weights (and usually inference configs/tokenizer)",
            "Sometimes partial code (e.g., inference scripts) but not full training stack"
        ],
        "what_you_can_do": [
            "Run inference locally/cloud; fine-tune / do domain adaptation on top of the weights",
            "Quantize, prune, distill, and package for different hardware targets"
        ],
        "what_you_cannot_do": [
            "Fully audit the original training process/data provenance end-to-end",
            "Reliably reproduce the original pretraining without full recipes/data"
        ],
    },
    "Architecture Only": {
        "definition": (
            "A model release where the architecture (paper/spec) is public, but pretrained weights are not provided, "
            "so you only know the design, not the learned parameters."
        ),
        "what_is_open": [
            "Architecture/specification (paper, diagrams, hyperparameters)",
            "Sometimes reference implementation code without released pretrained weights"
        ],
        "what_you_can_do": [
            "Implement the architecture and train your own weights if you have data + compute",
            "Study design choices and build compatible systems (tokenizers, prompts, etc.)"
        ],
        "what_you_cannot_do": [
            "Use the original model capabilities without training your own weights",
            "Claim comparable performance without significant training resources"
        ],
    },
}
open_source_levels
```

```python
# TODO: fill the comparison table (keep markdown formatting)
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Source code + architecture + pretrained weights (often configs/tokenizer too) | Maximum flexibility: you can audit, modify, fine-tune, and potentially retrain end-to-end; strongest option for compliance audits and long-term maintenance. |
| Weights Released | Pretrained weights (usually with inference configs/tokenizer; training stack may be partial) | Practical for domain fine-tuning and deployment; harder to fully audit training provenance, and you may be limited in changing core training behavior. |
| Architecture Only | Architecture/spec (sometimes reference code) but no pretrained weights | You cannot domain-adapt the original model; you must train from scratch to get usable weights, which is expensive and slow for most teams. |"""
print(comparison_table)
```

```python
# TODO: 3?5 sentence comparative paragraph
comparative_paragraph = (
    "Fully Open releases provide both the source code and pretrained weights, enabling deep inspection, modification, and end-to-end retraining when needed. "
    "Weights Released models are usually the most practical choice for teams because you can run and fine-tune the model, but you may not be able to fully audit or reproduce the original training pipeline. "
    "Architecture Only releases publish the design but not the learned parameters, which means you cannot use the original model without training your own weights. "
    "In practice, the more that is released (code + weights), the easier it is to debug, comply, and maintain the system over time."
)
print(comparative_paragraph)
```

## Exercise 4: Local Readiness Audit
**Use:** RAM/disk/OS; llama.cpp requirements (AVX/SSE, cmake/make/gcc/clang); quantized formats (GGUF).

**Step-by-step**
1) Gather system specs (RAM, free disk, OS type/version).  
2) Fill the audit table and mark ?/?.  
3) Check llama.cpp readiness (instruction sets, compilers, build tools).  
4) Identify upgrade needs for each ?.

**Deliverables:** filled table + upgrade summary.

```python
# TODO: your measured system specs (auto-detect where possible)
import os
import platform
import shutil

def get_ram_gb():
    # Linux / Colab
    try:
        import psutil
        return round(psutil.virtual_memory().total / (1024**3), 2)
    except Exception:
        pass

    # Fallback: /proc/meminfo
    try:
        with open("/proc/meminfo", "r") as f:
            for line in f:
                if line.startswith("MemTotal:"):
                    kb = int(line.split()[1])
                    return round(kb * 1024 / (1024**3), 2)
    except Exception:
        return None

def get_free_disk_gb(path="."):
    try:
        total, used, free = shutil.disk_usage(path)
        return round(free / (1024**3), 2)
    except Exception:
        return None

system_specs = {
    "ram_gb": get_ram_gb(),
    "free_disk_gb": get_free_disk_gb("."),
    "os": f"{platform.system()} {platform.release()} ({platform.machine()})",
}
system_specs
```

```python
# TODO: readiness table (keep markdown formatting)
ram_ok = system_specs["ram_gb"] is not None and system_specs["ram_gb"] >= 16
disk_ok = system_specs["free_disk_gb"] is not None and system_specs["free_disk_gb"] >= 40
os_ok = platform.system().lower() in ["linux"]  # WSL2 also reports Linux here

readiness_table = f"""| Requirement | Your System Specs | Meets Requirement? |
| --- | --- | --- |
| RAM (>= 16 GB) | {system_specs['ram_gb']} GB | {'✅' if ram_ok else '❌'} |
| Free Disk Space (>= 40 GB) | {system_specs['free_disk_gb']} GB | {'✅' if disk_ok else '❌'} |
| OS (Linux/WSL2) | {system_specs['os']} | {'✅' if os_ok else '❌'} |"""
print(readiness_table)
```
