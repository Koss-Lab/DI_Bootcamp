## Exercise 1: Rewrite and Optimize a Vague Prompt
**Scenario**: Manager asks for a LinkedIn post about FlowNest with the vague prompt `"Write something about productivity tips."`

### Tasks
- Identify 3 critical issues with the prompt.
- Rewrite it using: role prompting, clear instructions, and format/tone/length constraints.
- Add a final line to make it reusable for future social posts.

```python
# TODO: list three critical issues with the vague prompt
critical_issues = [
    "The prompt does not specify the target audience or the LinkedIn platform context.",
    "There is no brand voice or mention of the FlowNest product or its value proposition.",
    "No constraints on tone, format, or length, resulting in generic and unfocused output."
]
critical_issues

```

```python
# TODO: rewrite the prompt using role, instructions, and constraints
rewritten_prompt = (
    "## Role "
    "Act as a B2B SaaS marketing copywriter. "
    "## Task "
    "Write a LinkedIn post promoting FlowNest, a focus and productivity app designed for busy professionals. "
    "## Instructions "
    "Provide 3 actionable productivity tips that naturally highlight how FlowNest improves focus and workflow. "
    "## Style and Constraints "
    "Use a professional yet friendly tone, bullet points, and keep the post under 120 words. "
    "## Reusability "
    "Reuse this exact structure whenever creating social media posts for productivity tools."
)
rewritten_prompt
```

**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

```python
# TODO: write the full multi-part prompt
quiz_prompt = (
    "## Role "
    "You are an experienced middle school science teacher. "
    "## Audience "
    "7th grade students aged 11 to 13 years old. "
    "## Task "
    "Using the article provided below, generate educational content. "
    "## Output Requirements "
    "Create exactly 2 bullet points summarizing the article. "
    "Create 3 multiple choice questions, each with 1 correct answer and 2 plausible distractors. "
    "## Tone and Vocabulary "
    "Use a friendly, encouraging tone and simple vocabulary appropriate for ages 11 to 13. "
    "## Format "
    "Structure the output so it can be copied directly into Google Slides. "
    "## Source "
    "Article content will be provided below."
)
quiz_prompt
```

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
# TODO: baseline anti-hallucination prompt
no_hallucination_prompt = (
    "## Instruction "
    "Summarize the article using only information explicitly stated in the provided text. "
    "Do not add facts statistics projections or assumptions that are not present in the article."
)
no_hallucination_prompt
```

```python
# TODO: verification-focused rewrite
verification_prompt = (
    "## Verification Rules "
    "Summarize the article strictly using the provided text. "
    "Only include claims that are explicitly stated in the article. "
    "If specific data or conclusions are not mentioned clearly state that the information is not available."
)
verification_prompt
```

```python
# TODO: list mitigation strategies and risky domains
mitigation_strategies = [
    "Restricting the model to the provided source text only.",
    "Explicitly instructing the model to acknowledge missing information instead of guessing."
]
mitigation_strategies

risky_domains = {
    "domain_1": {
        "name": "Healthcare",
        "why": "Hallucinated medical information can lead to harmful diagnoses or treatments."
    },
    "domain_2": {
        "name": "Legal",
        "why": "Fabricated laws or precedents can cause serious legal and financial consequences."
    }
}
risky_domains
```
