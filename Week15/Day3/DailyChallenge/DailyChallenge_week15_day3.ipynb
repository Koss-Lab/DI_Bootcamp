# Exercise 1 — Understanding LLM Evaluation

### 1.1 Why evaluating LLMs is more complex than traditional software
Evaluating LLMs is more complex because:
- Their outputs are **probabilistic**, not deterministic.
- There is **no single correct answer** for most natural language tasks.
- Their behavior depends heavily on **context, prompt phrasing, and randomness**.
- They may produce **plausible but incorrect** outputs (hallucinations).
- Measuring quality requires **semantic understanding**, which is difficult to automate.

### 1.2 Key reasons for evaluating an LLM’s safety
We evaluate safety to prevent:
- **Harmful outputs** (hate speech, toxicity).
- **Factual inaccuracies** leading to real-world consequences.
- **Bias propagation** from training data.
- **Dangerous instructions** (self-harm, illegal activity).
- **Privacy violations** (leaking personal data).

### 1.3 How adversarial testing contributes to LLM improvement
Adversarial testing:
- Exposes model weaknesses by using **tricky or misleading prompts**.
- Reveals how the model behaves under **edge cases** or **ambiguous phrasing**.
- Helps developers improve **robustness, safety alignment, and error detection**.

### 1.4 Limitations of automated evaluation metrics vs. human evaluation
Automated metrics:
- Can measure **surface similarity**, not true meaning.
- Fail when evaluating **style, creativity, nuance, or reasoning**.
- Can be gamed by producing **overly literal or extractive summaries**.

Human evaluation:
- Captures **semantic meaning**, tone, factuality, and quality.
- But is **slow, expensive, and subjective**.

A complete evaluation requires **both automated metrics and human judgment**.


# Exercise 2 — Applying BLEU and ROUGE Metrics
Below we compute BLEU and ROUGE for the given examples.


```python
#!pip install rouge_score==0.1.2
#!pip install nltk
#!pip install evaluate
```

```python
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

reference_bleu = "Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation."
generated_bleu = "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application."

ref_tokens = [reference_bleu.split()]
gen_tokens = generated_bleu.split()

smooth = SmoothingFunction().method1

bleu_score_value = sentence_bleu(ref_tokens, gen_tokens, smoothing_function=smooth)
bleu_score_value
```

```python
import evaluate

rouge = evaluate.load("rouge")

reference_rouge = [
    "In the face of rapid climate change, global initiatives must focus on reducing carbon emissions and developing sustainable energy sources to mitigate environmental impact."
]

generated_rouge = [
    "To counteract climate change, worldwide efforts should aim to lower carbon emissions and enhance renewable energy development."
]

scores = rouge.compute(predictions=generated_rouge, references=reference_rouge)
scores
```

### BLEU and ROUGE Limitations
Both metrics focus on **n-gram overlap**, which causes problems:

- They penalize **paraphrasing**, even when meaning is identical.
- They fail on **creative, narrative, or stylistic** tasks.
- They cannot judge **factual correctness**, **reasoning**, or **coherence**.
- High scores do not guarantee usefulness; low scores may still be good outputs.

### Suggested Improvements
Better evaluation approaches include:
- **BERTScore**: uses contextual embeddings.
- **GPT-based evaluators**: model-graded evaluation.
- **Human scoring** for fluency, coherence, factual accuracy.
- **Task-specific evaluation** (e.g., QA accuracy, factual consistency checking).


# Exercise 3 — Perplexity Analysis

### 3.1 Which model has lower perplexity?
Perplexity = 1 / P(word).

- Model A probability: 0.8 → perplexity = 1.25  
- Model B probability: 0.4 → perplexity = 2.5  

**Model A has lower perplexity**, meaning it is more confident and better fits the data.

### 3.2 Meaning of perplexity = 100
A perplexity of 100 means:
- The model is **very unsure** about upcoming words.
- The model likely produces **poor-quality, inconsistent, or incoherent text**.

### Improving perplexity:
- Increase model size.
- Train with better data.
- Fine-tune on the target domain.
- Use better tokenization or normalization.
- Reduce noise in training data.

# Exercise 6 — Comparative Analysis of Evaluation Methods

### Chosen task: **Text Summarization**

### Metric Comparison
| Metric | Strengths | Weaknesses |
|--------|-----------|------------|
| **ROUGE** | Good for extractive summarization, measures recall, widely used | Fails for paraphrasing, ignores semantics |
| **BERTScore** | Embedding-based, captures meaning, robust to paraphrasing | More computationally expensive |
| **Human Evaluation** | Best for fluency, coherence, factuality | Slow, costly, subjective |
| **Perplexity** | Measures model fit to language | Not correlated with summarization quality |

### Most appropriate metric
**BERTScore + Human Evaluation** is ideal because:
- Summaries require **semantic understanding**, not n-gram overlap.
- Human evaluation ensures **coherence, factual accuracy, and readability**.
- ROUGE can still be included but is insufficient alone.
