## 🌟 Exercise 1: Traditional vs. Modern NLP: Comparative Analysis
1) Complete the table (replace each TODO):

| Aspect | Traditional NLP | Modern NLP |
|---|---|---|
| Feature Engineering | Manual feature engineering: hand-crafted features such as bag-of-words, n-grams, TF-IDF, POS tags, dependency features. | Automatic feature learning: deep models learn features directly from raw text (end-to-end representation learning). |

| Advantages | Simple, relatively interpretable, cheap to train, can work with smaller datasets and modest hardware. | Much higher accuracy, strong ability to capture long-range dependencies and semantics, easy transfer to many tasks, better robustness to noise and domain variety. |
| Disadvantages | Heavy reliance on manual feature engineering, brittle to domain shift, poor handling of complex context and long documents. | Computationally expensive, requires large datasets and hardware, less interpretable, can encode and amplify biases from pre-training data. |

2) Discuss: How did the shift to modern NLP impact scalability and efficiency?

- TODO: Write 5-7 sentences connecting representation learning, hardware parallelism, and transfer learning.

The shift to modern NLP made it possible to scale models across many tasks with a single pre-trained backbone. Instead of engineering features and training a separate model for each problem, we now reuse the same large transformer and adapt it with light fine-tuning or even just prompting. This greatly improves developer efficiency, because most of the heavy computation is done once during pre-training. Modern architectures also map well to GPU/TPU hardware thanks to parallelizable attention operations, which improves throughput compared to many older sequential methods. At the same time, the cost of training and serving very large models increased, so scalability in terms of model size and deployment became a new challenge. Overall, modern NLP trades manual feature engineering and many small models for a few very large, reusable models that are efficient at scale but expensive individually.

### BERT
- TODO: Bidirectional encoder-only; MLM (and historically NSP) : Bidirectional encoder-only transformer trained with masked language modeling (and originally next sentence prediction). It attends to the full left and right context simultaneously when encoding each token.
- Application: TODO (e.g., classification/NER/Q&A) : Text classification and token-level tasks such as sentiment analysis, NER, and extractive question answering.
- Why: TODO : Because BERT builds rich contextual embeddings for every token using both left and right context, it produces strong sentence and token representations that are ideal for classification heads or span selection layers. It does not need to generate long sequences; it mainly needs accurate understanding of the input, which is exactly what an encoder-only model is optimized for.

### GPT
- TODO: Decoder-only; autoregressive causal LM : Decoder-only, autoregressive causal language model. Each token only attends to previous tokens (left-to-right), optimized to predict the next token.
- Application: TODO (e.g., chat/generation/code) : Open-ended text generation such as chatbots, story generation, code completion, and general dialogue systems.
- Why: TODO : GPT’s causal formulation directly matches the generation process: it models the probability of each next token given the history, which leads to fluent and coherent text continuation. Decoder-only transformers are easy to sample from and can flexibly generate variable-length sequences, making them a natural choice for conversational and generative applications.

### T5
- TODO: Encoder-decoder; text-to-text framework : Encoder-decoder transformer using a unified “text-to-text” framework: every task is cast as mapping an input text sequence to an output text sequence.
- Application: TODO (e.g., summarization/translation) : Sequence-to-sequence tasks like abstractive summarization, machine translation, question answering, and text rewriting.
- Why: TODO : T5’s encoder can fully absorb and compress the input sequence, while the decoder conditions on this representation to generate a structured output. This separation between understanding (encoder) and generation (decoder) is ideal for tasks where the output is a transformed version of the input, not just a continuation.


## 🌟 Exercise 3: Benefits and Ethics of Pre-training
- Benefits (explain each in your words):
  1) Improved generalization: TODO : Pre-trained models learn broad linguistic and world knowledge from massive unlabeled corpora, so they capture patterns that go far beyond any single supervised dataset. As a result, they generalize better to new tasks, domains, or phrasings compared to models trained only on small, task-specific data.
  2) Less labeled data: TODO : Because much of the representation learning is done during unsupervised pre-training, downstream tasks can reach strong performance with comparatively small labeled datasets. Fine-tuning often requires only thousands or even hundreds of examples instead of millions.
  3) Faster fine-tuning: TODO : Fine-tuning a pre-trained model is far cheaper than training from scratch. Only the last few layers or a small number of adapter parameters need to be updated, which reduces both training time and energy consumption.
  4) Transfer learning: TODO : Knowledge learned on one distribution (e.g., general web text) can be transferred to many other domains like legal, medical, or customer support. This transfer learning enables rapid prototyping of new NLP applications without starting over each time.
  5) Robustness: TODO : Pre-trained models see a wide variety of language, styles, and topics, which often makes them more robust to noise, typos, and out-of-distribution phrasing than narrowly trained models. They can handle edge cases that were never explicitly annotated.

- Ethical concerns: TODO (bias, misinformation, misuse, privacy) : Pre-training on massive web-scale datasets inevitably exposes models to biased, offensive, or misleading content. The model can internalize stereotypes about gender, race, religion, or geography and reproduce them in its outputs. Large models can also be misused to generate convincing misinformation, spam, or deepfake text at scale. Privacy is another concern, since training data may accidentally include personal or sensitive information that the model could later regurgitate.
- Mitigations: TODO (data curation, differential privacy, safety filters, RLHF, audits) : Mitigation strategies include careful data curation and filtering to remove toxic or highly biased sources, as well as balancing datasets across demographic groups. Techniques such as differential privacy can limit memorization of specific user data. Safety filters and post-processing classifiers can detect and block harmful generations at inference time. Reinforcement learning from human feedback (RLHF) or similar preference-based tuning can steer models toward safer, more helpful behavior. Regular audits and red-teaming exercises help surface remaining failure modes so that mitigations can be iteratively improved.


## 🌟 Exercise 4: Transformer Architecture Deep Dive
### Self-Attention & Multi-Head Attention
- TODO: Describe the Q/K/V flow, softmax weighting, and value mixing. : In self-attention, each token embedding is linearly projected into three vectors: a query (Q), a key (K), and a value (V). For a given token, we compute dot-products between its query and all keys to measure how relevant each other token is. These scores are scaled and passed through a softmax to produce attention weights that sum to 1. The output for that token is then a weighted sum of all value vectors, where more relevant tokens contribute more. This mechanism lets each position aggregate information from any other position in the sequence.
- TODO: Explain why multiple heads help (subspace projections, diverse relations). : Multi-head attention duplicates this process several times in parallel with different learned projections for Q, K, and V. Each head operates in a different subspace of the embedding, encouraging the model to learn diverse relationships (e.g., syntactic dependencies, coreference, topic, or position). The outputs of all heads are concatenated and linearly mixed, giving the model a richer and more expressive representation than a single attention head could provide.
- Example sentence (different from lesson):
  - TODO: Provide a sentence and describe at least two different relations distinct heads might capture. : Consider the sentence: “The engineer who fixed the server presented her solution yesterday.” One attention head might focus on grammatical structure, strongly connecting “engineer” with “presented” to model the subject–verb relationship. Another head might focus on coreference, linking “engineer” to “her” so that pronoun resolution is encoded. Additional heads could emphasize temporal information (“yesterday”) or objects (“fixed” ↔ “server”, “presented” ↔ “solution”).

### Pre-training Objectives
- MLM vs CLM: TODO (compare objectives and typical use) : Masked Language Modeling (MLM) randomly masks some tokens in the input and trains the model to predict the original tokens using both left and right context. It is typically used in encoder-only models like BERT. Causal Language Modeling (CLM) trains the model to predict the next token given only the previous tokens, enforcing a left-to-right constraint; this is standard for decoder-only models such as GPT.
- When to prefer MLM vs CLM: TODO : MLM is preferable when the goal is to learn strong bidirectional representations for understanding tasks like classification, NER, or extractive QA. CLM is preferable for generative tasks where we need to sample coherent sequences token by token, such as story writing, dialogue, or code completion.
- NSP: TODO (why early BERT used it and why many modern models avoid it) : Early BERT models used Next Sentence Prediction (NSP) to help the model understand sentence-level relationships by classifying whether two segments appeared consecutively in the original text. Later research showed that NSP contributed limited gains and could even hurt performance, while adding complexity. Many modern models replace NSP with better objectives (e.g., sentence order prediction or purely span-level tasks) or drop it entirely in favor of more focused language modeling.

### Transformer Model Selection
- Sentiment on reviews: TODO (Encoder-only / Decoder-only / Encoder-Decoder) + justification : Encoder-only model (BERT-style). For sentiment analysis, the main requirement is a high-quality representation of the input text and a classification head; no long output sequence needs to be generated. Encoder-only transformers are efficient and very strong for such understanding tasks.
- Conversational chatbot (creative responses): TODO + justification : Decoder-only model (GPT-style). A chatbot must generate flexible, varied responses conditioned on the conversation history. Decoder-only causal LMs are optimized for next-token prediction and long-form generation, making them ideal for creative and interactive dialogue.
- Technical document translation (EN→ES): TODO + justification : Encoder-decoder model (T5- or classic seq2seq-style). The encoder can fully process and disambiguate the source sentence, while the decoder generates the target language conditioned on this representation. This architecture aligns naturally with translation, especially when the output needs to be closely aligned in meaning and structure to the input.

### Positional Encoding
- Purpose: TODO (why absolute/relative position info matters) : Self-attention is permutation-invariant: without additional information, it treats the input as an unordered set of tokens. Positional encoding injects information about token order (absolute or relative positions) into the embeddings so that the model can distinguish “dog bites man” from “man bites dog.” It allows the transformer to model word order, syntax, and long-range sequential patterns.
- Example failure without positions: TODO : Without positional encoding, the sentences “She ate the cake before the meeting” and “Before the meeting she ate the cake” would look almost identical to the model, even though the order conveys subtle differences in emphasis and structure. More critically, time-ordered instructions like “First, log in. Then, change your password.” could be interpreted the same as “Change your password, then log in,” which is semantically wrong.

## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator
### 1) Temperature Scenarios
- T=0.2: TODO : Very low temperature makes the distribution sharp, heavily favoring the most probable tokens. Outputs become more deterministic, conservative, and repetitive, with high coherence but limited creativity.
- T=1.5: TODO : High temperature flattens the distribution, giving more weight to low-probability tokens. The model produces more surprising and diverse text, but coherence and factual accuracy degrade; outputs may become rambling or off-topic.
- T=1.0: TODO : Temperature 1.0 corresponds to the “natural” model distribution. It balances diversity and coherence, often used as a reasonable default for many applications.

### 2) Application Design
- Bedtime stories (creativity vs coherence): TODO (temperature strategy and why) : For personalized bedtime stories, I would use a moderate to slightly high temperature (e.g., 0.8–1.2) to keep the stories imaginative and varied while still understandable. I might start the story with a lower temperature to establish a clear setup, then increase it slightly in the middle for creative plot twists, and lower it again near the end to ensure a stable, calming conclusion. Additional constraints (e.g., safety filters and maximum length) would keep the content appropriate for children.
- Financial report summaries (accuracy/reliability): TODO (temperature/decoding strategy and why) : For financial summaries, I would use a low temperature (around 0.2–0.5) and possibly deterministic decoding like beam search or greedy decoding. The goal is to produce accurate, concise, and consistent text that closely reflects the input data, not to be creative. Lower temperature reduces the chance of hallucinated numbers or speculative language, improving reliability and repeatability.

### 3) Temperature & Bias
- TODO: How does temperature influence bias surfacing or dampening? Give a realistic example. : Temperature interacts with bias by changing how likely rare but potentially problematic tokens are to appear. At very low temperature, the model mostly outputs the most common, high-probability patterns, which can reinforce majority stereotypes (e.g., repeatedly associating certain professions with one gender). At very high temperature, the model samples more from the tail of the distribution, which can surface unusual or toxic associations that are rarely seen but still present in the model. For example, a hiring-assistant chatbot with high temperature might occasionally generate overtly biased suggestions, while a very low temperature might consistently favor one demographic over others in subtle ways. Careful temperature choice plus bias mitigation techniques are needed in sensitive applications.


### (Optional) Quick Generation Demo Across Temperatures
Note: Requires `transformers` and model download; skip if offline.


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

# Filled TODOs
model_name = "gpt2"  # causal LM suitable for temperature demo
prompt = "The future of artificial intelligence depends on"  # custom prompt

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print("\n--- Temperature:", T)
    print(out[0]["generated_text"])  # show how style changes with temperature
```
