```python
# Cell 0 — Setup
import numpy as np
np.set_printoptions(suppress=True, precision=2)  # nicer printing
np.random.seed(42)  # reproducibility
print("✅ NumPy ready:", np.__version__)
```

```python
# Cell 1 — Exercise 1: 5x5 random array + min/max
arr_5x5 = np.random.rand(5, 5)  # uniform in [0, 1)
print("5x5 random array:\n", arr_5x5)
print("Min:", round(arr_5x5.min(), 2), "Max:", round(arr_5x5.max(), 2))
```

```python
# Cell 2 — Exercise 2: Normalize a 3x3 random matrix
M = np.random.randn(3, 3)  # standard normal entries
mu = M.mean()
sigma = M.std()
M_norm = (M - mu) / sigma

print("Original 3x3 matrix:\n", M)
print("\nNormalized 3x3 matrix:\n", M_norm)

# sanity check: approx mean 0, std 1
print("\nCheck ~0 mean, ~1 std:",
      round(M_norm.mean(), 6), round(M_norm.std(), 6))
```

```python
# Cell 4 — Exercise 4: (5x3) @ (3x2)
A = np.arange(1, 1 + 5*3, dtype=float).reshape(5, 3)     # 1..15 → 5x3
B = (np.arange(1, 1 + 3*2, dtype=float).reshape(3, 2) )  # 1..6  → 3x2

C = A @ B  # same as np.dot(A, B)

print("A (5x3):\n", A)
print("\nB (3x2):\n", B)
print("\nA @ B (5x2):\n", C)
```

```python
# Cell 5 — Exercise 5: Create two compatible matrices and multiply
# Example: (2x4) @ (4x3) -> (2x3)
A2 = np.array([[2, 0, -1, 3],
               [1, 4,  2, 0]], dtype=float)  # shape (2,4)

B2 = np.array([[ 1,  2, 0],
               [-1,  3, 1],
               [ 4, -2, 2],
               [ 0,  5, 1]], dtype=float)     # shape (4,3)

C2 = A2 @ B2

print("A2 shape:", A2.shape, "\n", A2)
print("\nB2 shape:", B2.shape, "\n", B2)
print("\nA2 @ B2 shape:", C2.shape, "\n", C2)
```
