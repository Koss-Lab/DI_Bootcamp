```python
# Cell 0 — Setup
import numpy as np
print("✅ NumPy ready, version:", np.__version__)
```

```python
# Cell 1 — Exercise 1: Array Creation and Manipulation
arr1 = np.arange(10)
print(arr1)
```

```python
# Cell 2 — Exercise 2: Type Conversion and Array Operations
lst = [3.14, 2.17, 0, 1, 2]
arr2 = np.array(lst, dtype=int)
print(arr2)
```

```python
# Cell 3 — Exercise 3: Working with Multi-Dimensional Arrays
arr3 = np.arange(1, 10).reshape(3, 3)
print(arr3)
```

```python
# Cell 4 — Exercise 4: Creating Multi-Dimensional Array with Random Numbers
arr4 = np.random.rand(4, 5)  # values between 0 and 1
print(arr4)
```

```python
# Cell 5 — Exercise 5: Indexing Arrays
arr5 = np.array([[21,22,23,22,22],
                 [20,21,22,23,24],
                 [21,22,23,22,22]])
second_row = arr5[1]   # index 1 = second row
print(second_row)
```

```python
# Cell 6 — Exercise 6: Reversing elements
arr6 = np.arange(10)
reversed_arr6 = arr6[::-1]
print(reversed_arr6)
```

```python
# Cell 7 — Exercise 7: Identity Matrix
arr7 = np.eye(4)
print(arr7)
```

```python
# Cell 8 — Exercise 8: Sum and Average
arr8 = np.arange(10)
total = arr8.sum()
average = arr8.mean()
print("Sum:", total, "Average:", average)
```

```python
# Cell 9 — Exercise 9: Create Array and Reshape
arr9 = np.arange(1, 21).reshape(4, 5)
print(arr9)
```

```python
# Cell 10 — Exercise 10: Conditional Selection (odd numbers)
arr10 = np.arange(10)
odd_numbers = arr10[arr10 % 2 == 1]
print(odd_numbers)
```
