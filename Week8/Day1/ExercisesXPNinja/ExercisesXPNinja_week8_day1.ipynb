```python
# Cell 0 — Setup
import numpy as np
np.set_printoptions(precision=2, suppress=True)  # nicer printing
np.random.seed(42)  # reproducibility
print("✅ NumPy ready:", np.__version__)
```

```python
# Cell 1 — Exercise 1: Create 5x5 random matrix and replace max with 0
A = np.random.rand(5, 5)  # values in [0,1)
print("Original 5x5 matrix:\n", A)

# Find index of the maximum value
max_idx = np.unravel_index(np.argmax(A), A.shape)

# Replace max with 0 (make a copy to keep original if needed)
A_replaced = A.copy()
A_replaced[max_idx] = 0.0

print("\nIndex of max:", max_idx, " | max value:", float(A[max_idx]))
print("\nMatrix after replacing max with 0:\n", A_replaced)
```

```python
# Cell 2 — Exercise 2: Common values between two random arrays of size 5
# Using rounding to 2 decimals so intersections are more likely to appear
x = np.random.rand(5)
y = np.random.rand(5)

x_round = np.round(x, 2)
y_round = np.round(y, 2)

common = np.intersect1d(x_round, y_round)

print("x (rounded):", x_round)
print("y (rounded):", y_round)
print("Common values:", common)
```

```python
# Cell 3 — Exercise 3: Sort ascending then descending
arr = np.random.rand(10)
asc = np.sort(arr)
desc = asc[::-1]

print("Original:", np.round(arr, 2))
print("Ascending:", np.round(asc, 2))
print("Descending:", np.round(desc, 2))
```

```python
# Cell 4 — Exercise 4: Random 4x4 matrix rank
M = np.random.randn(4, 4)
rank = np.linalg.matrix_rank(M)

print("Matrix 4x4:\n", M)
print("\nRank:", rank)
```

```python
# Cell 5 — Exercise 5: 2D array with 1s on border and 0s inside
n = 5  # size per the example
border = np.ones((n, n), dtype=int)
border[1:-1, 1:-1] = 0
print(border)
```
