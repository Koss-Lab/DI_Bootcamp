```python
# Cell 1 — Exercise 1: determinant & inverse of a 3x3 matrix
A = np.array([[2., 1., 3.],
              [0., 1., 4.],
              [5., 2., 1.]])

det_A = np.linalg.det(A)
print("Matrix A:\n", A)
print("\nDeterminant(A) =", det_A)

if np.isclose(det_A, 0.0):
    print("\n❌ Matrix A is singular — no inverse.")
else:
    A_inv = np.linalg.inv(A)
    print("\nInverse(A):\n", A_inv)
    # sanity check: A @ A_inv ≈ I
    I = A @ A_inv
    print("\nA @ A_inv (should be Identity):\n", I)
```

```python
# Cell 6 — Exercise 6: paired test for training effectiveness (NumPy-only)

productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after  = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Hypothesis:
# H0: mean(after - before) = 0  (no improvement)
# H1: mean(after - before) > 0  (improvement)
diff = productivity_after - productivity_before
n = diff.size
mean_diff = diff.mean()
std_diff = diff.std(ddof=1)  # sample std
t_stat = mean_diff / (std_diff / np.sqrt(n))

# For a quick NumPy-only decision, compare to critical t (~2.045 for df=29, α=0.05 one-sided ~1.699)
t_crit_one_sided_5pct = 1.699  # approximate
decision = "Reject H0 (evidence of improvement)" if t_stat > t_crit_one_sided_5pct else "Fail to reject H0"

print(f"mean(before) = {productivity_before.mean():.2f}")
print(f"mean(after)  = {productivity_after.mean():.2f}")
print(f"mean(after - before) = {mean_diff:.2f}")
print(f"t-statistic = {t_stat:.3f}  | threshold ≈ {t_crit_one_sided_5pct}")
print("Decision:", decision)
```

```python
# Cell 10 — Exercise 10: simple line plot of random numbers
x = np.arange(0, 50)
y = np.cumsum(np.random.randn(50))  # random walk

plt.figure(figsize=(8,4))
plt.plot(x, y, marker="o", linewidth=1)
plt.title("Random Walk (NumPy) — Line Plot (Matplotlib)")
plt.xlabel("Step")
plt.ylabel("Value")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
