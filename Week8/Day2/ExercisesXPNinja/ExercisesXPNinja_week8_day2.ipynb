```python
# Cell 0 — Setup
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=3, suppress=True)
np.random.seed(42)  # reproducibility

print("✅ NumPy & Matplotlib ready")
```

```python
# Cell 1 — Extract diagonal elements and build diagonal matrix
M = np.arange(1, 17).reshape(4, 4)
print("Original 4x4 matrix:\n", M)

# Extract diagonal
diag = np.diag(M)
print("\nDiagonal elements:", diag)

# Build diagonal matrix from these elements
D = np.diag(diag)
print("\nDiagonal matrix:\n", D)
```

```python
# Cell 2 — Replace values >0.5 with -1
arr = np.random.rand(20)
print("Original array:\n", np.round(arr, 3))

arr_cond = arr.copy()
arr_cond[arr_cond > 0.5] = -1
print("\nArray after conditional replacement:\n", np.round(arr_cond, 3))
```

```python
# Cell 3 — Normalize array (z-score)
arr2 = np.random.rand(10)
mean = arr2.mean()
std = arr2.std(ddof=0)

normalized = (arr2 - mean) / std
print("Original:", np.round(arr2, 3))
print("Mean:", mean, "Std:", std)
print("Normalized:", np.round(normalized, 3))
```

```python
# Cell 4 — Pearson correlation using NumPy
x = np.random.rand(15)
y = np.random.rand(15)

# Formula implementation
num = np.sum((x - x.mean()) * (y - y.mean()))
den = np.sqrt(np.sum((x - x.mean())**2) * np.sum((y - y.mean())**2))
r = num / den

print("x:", np.round(x, 3))
print("y:", np.round(y, 3))
print("Pearson correlation r:", round(r, 3))

# Cross-check with numpy.corrcoef
print("np.corrcoef result:", np.corrcoef(x, y)[0,1])
```

```python
# Cell 5 — Time series with linear trend fit
days = np.arange(1, 16)
values = np.random.randint(50, 150, size=15)

# Fit linear regression (polyfit degree=1)
a, b = np.polyfit(days, values, 1)
trend = a * days + b

print("Values:", values)
print(f"Trend line: y = {a:.2f}*x + {b:.2f}")

# Plot
plt.figure(figsize=(8,4))
plt.plot(days, values, "o-", label="Observed")
plt.plot(days, trend, "r--", label="Trend line")
plt.title("Time Series with Linear Trend")
plt.xlabel("Day")
plt.ylabel("Value")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
