# Exercise 1: Identifying Limitations of Traditional Language Models

## Sentence Analysis
**Sentence:** "The scientist, who had been working on the project for years, finally made a breakthrough discovery."

## How Traditional Models Struggle

### Sequential Processing Challenge
Traditional language models (like RNNs and LSTMs) process text sequentially, word by word:
- They read: "The" → "scientist" → "who" → "had" → "been" → ... → "discovery"
- By the time the model reaches "discovery," the information about "scientist" (13 words earlier) may have degraded in the model's memory
- The intervening clause "who had been working on the project for years" creates a long-range dependency that traditional models struggle to maintain

### Memory Degradation
- **Vanishing Gradient Problem**: In RNNs/LSTMs, information from early words weakens as the sequence progresses
- **Limited Context Window**: Traditional models have difficulty maintaining long-term dependencies beyond 10-20 words
- **Sequential Bottleneck**: Each word must pass through all previous hidden states, creating a bottleneck for information flow

## Impact on NLP Tasks

### Question Answering
- **Question**: "Who made the discovery?"
- **Problem**: The model might lose the connection between "scientist" and "discovery," potentially answering incorrectly or vaguely

### Summarization
- **Problem**: The model might fail to capture the key relationship (scientist → breakthrough discovery) and produce incomplete summaries
- **Example**: Might summarize as "Someone made a discovery" instead of "The scientist made a breakthrough discovery after years of work"

### Semantic Understanding
- The model might not understand that:
  - The scientist is the subject performing the action
  - Years of work led to the discovery
  - "Breakthrough" modifies "discovery"

## How Attention Mechanisms Solve This

### Direct Access to All Positions
- **Attention allows the model to look at ALL words simultaneously**, regardless of distance
- When processing "discovery," attention can directly focus on "scientist" without degradation
- No sequential bottleneck—direct connections between any two words

### Weighted Relevance
```
When processing "discovery," attention weights might look like:
- "scientist": 0.45 (high attention - subject of action)
- "breakthrough": 0.30 (high attention - modifier)
- "made": 0.15 (moderate attention - main verb)
- "the", "who", "been": 0.02 each (low attention - less relevant)
```

### Multi-Head Attention
- Different attention heads can capture different relationships:
  - Head 1: scientist ↔ discovery (subject-object)
  - Head 2: breakthrough ↔ discovery (adjective-noun)
  - Head 3: working ↔ years (duration)

### Benefits for NLP Tasks
- **Question Answering**: Model accurately links "scientist" to "discovery"
- **Summarization**: Captures complete relationships in concise form
- **Translation**: Maintains grammatical relationships across languages
- **Semantic Understanding**: Full comprehension of sentence structure and meaning

## Conclusion
Attention mechanisms revolutionized NLP by eliminating the sequential processing limitation. Models can now capture long-range dependencies, understand complex contexts, and maintain relationships between distant words—all crucial for accurate language understanding and generation.

# Exercise 2: Impact of Attention in Transformers

## Chosen Task: Machine Translation (English → French)

## How Transformers Use Attention in Translation

### Self-Attention in Encoder
The encoder uses self-attention to understand the source sentence (English):

**Example Sentence:** "The bank can guarantee deposits will eventually cover future tuition costs because it is backed by the state."

**Attention helps resolve:**
1. **Word "bank"**: Is it a financial institution or a river bank?
   - Attention focuses on: "guarantee," "deposits," "backed" → Financial institution
2. **Word "it"**: What does "it" refer to?
   - Attention focuses on: "bank" (not "state" or "deposits")
3. **Phrase "cover future tuition costs"**: What's the subject?
   - Attention links: "deposits" → "will cover" → "costs"

### Cross-Attention in Decoder
The decoder uses cross-attention to align French words with English words:

**Translation Process:**
```
English: "The bank can guarantee deposits..."
French:  "La banque peut garantir les dépôts..."

When generating "banque" (bank):
- Cross-attention focuses heavily on "bank" (0.85 weight)
- Also considers "the" (0.10) for gender/article
- Minimal attention to other words

When generating "garantir" (guarantee):
- Cross-attention focuses on "guarantee" (0.90)
- Considers "can" (0.08) for tense/mood
```

### Handling Long-Range Dependencies

**Complex Sentence:**
"The company, which was founded in 1995 and has offices in 15 countries, announced quarterly results."

**Without Attention (Traditional Seq2Seq):**
- By the time the model processes "announced," it may have forgotten "company"
- Translation might lose the subject: "A annoncé des résultats" (missing subject)

**With Attention:**
- When generating the French subject, attention directly accesses "company"
- Correct translation: "L'entreprise... a annoncé des résultats trimestriels"
- Maintains relationship across 15+ words

````python
import wikipedia

# Set language to English
wikipedia.set_lang("en")

# Download article on Artificial Intelligence
print("Downloading Wikipedia article on 'Artificial Intelligence'...")
article = wikipedia.page("Artificial Intelligence")

# Extract content
knowledge_base = article.content

print(f"✅ Article downloaded!")
print(f"Title: {article.title}")
print(f"Length: {len(knowledge_base)} characters")
print(f"\nFirst 500 characters:\n{knowledge_base[:500]}...")
````

````python
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Split article into chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)

chunks = text_splitter.split_text(knowledge_base)

print(f"✅ Article split into {len(chunks)} chunks")
print(f"\nFirst chunk preview:\n{chunks[0][:300]}...")
````

````python
from sentence_transformers import SentenceTransformer
import numpy as np

# Load BERT model for embeddings
print("Loading BERT model for embeddings...")
bert_model = SentenceTransformer('all-MiniLM-L6-v2')

# Generate embeddings for all chunks
print("Generating embeddings for all chunks...")
chunk_embeddings = bert_model.encode(chunks, show_progress_bar=True)

print(f"✅ Embeddings generated!")
print(f"Embedding shape: {chunk_embeddings.shape}")
print(f"Dimension: {chunk_embeddings.shape[1]}")
````

````python
import faiss

# Create FAISS index
dimension = chunk_embeddings.shape[1]  # 384 for all-MiniLM-L6-v2
index = faiss.IndexFlatL2(dimension)

# Add embeddings to index
index.add(np.array(chunk_embeddings).astype('float32'))

print(f"✅ FAISS index created!")
print(f"Total vectors in index: {index.ntotal}")
````

````python
def retrieve_documents(question, k=4):
    """
    Retrieve top-k most relevant documents for a question

    Args:
        question: User's question
        k: Number of documents to retrieve

    Returns:
        List of relevant document chunks
    """
    # Encode the question using BERT
    question_embedding = bert_model.encode([question])

    # Search in FAISS index
    distances, indices = index.search(
        np.array(question_embedding).astype('float32'),
        k
    )

    # Get the chunks
    retrieved_docs = [chunks[i] for i in indices[0]]

    return retrieved_docs, distances[0]

# Test retrieval
test_question = "What is machine learning?"
docs, distances = retrieve_documents(test_question, k=3)

print(f"✅ Retriever function created!")
print(f"\nTest retrieval for: '{test_question}'")
print("="*80)
for i, (doc, dist) in enumerate(zip(docs, distances)):
    print(f"\nDocument {i+1} (distance: {dist:.4f}):")
    print(doc[:200] + "...")
    print("-"*80)
````

````python
import torch  # ← AJOUTE CETTE LIGNE
from transformers import GPT2LMHeadModel, GPT2Tokenizer, pipeline

# Load GPT-2 model and tokenizer
print("Loading GPT-2 model for generation...")
gpt_model_name = "gpt2"
gpt_tokenizer = GPT2Tokenizer.from_pretrained(gpt_model_name)
gpt_model = GPT2LMHeadModel.from_pretrained(gpt_model_name)

# Create text generation pipeline
generator = pipeline(
    'text-generation',
    model=gpt_model,
    tokenizer=gpt_tokenizer,
    device=0 if torch.cuda.is_available() else -1  # Use GPU if available
)

print("✅ GPT-2 generator loaded!")
````

````python
def rag_answer(question, k=4, max_length=200):
    """
    Complete RAG system: Retrieve relevant docs and generate answer

    Args:
        question: User's question
        k: Number of documents to retrieve
        max_length: Maximum length of generated answer

    Returns:
        Generated answer and retrieved documents
    """
    # Step 1: Retrieve relevant documents
    retrieved_docs, distances = retrieve_documents(question, k=k)

    # Step 2: Combine retrieved documents as context
    context = "\n\n".join(retrieved_docs)

    # Limit context length (GPT-2 has 1024 token limit)
    if len(context) > 2000:
        context = context[:2000]

    # Step 3: Create prompt for GPT-2
    prompt = f"""Based on the following information, answer the question.

Context:
{context}

Question: {question}

Answer:"""

    # Step 4: Generate answer
    generated = generator(
        prompt,
        max_length=len(prompt.split()) + max_length,
        num_return_sequences=1,
        temperature=0.7,
        top_p=0.9,
        do_sample=True,
        pad_token_id=gpt_tokenizer.eos_token_id
    )

    # Extract the answer (remove the prompt)
    full_text = generated[0]['generated_text']
    answer = full_text.split("Answer:")[-1].strip()

    # Clean up the answer (remove extra text after first complete sentence/paragraph)
    if '\n\n' in answer:
        answer = answer.split('\n\n')[0]

    return {
        'question': question,
        'answer': answer,
        'retrieved_docs': retrieved_docs,
        'distances': distances
    }

print("✅ RAG system ready!")
````

````python
# Test questions
test_questions = [
    "When was the term 'Artificial Intelligence' first coined?",
    "What is machine learning?",
    "How is AI used in healthcare?",
    "What are the risks of AI?"
]

print("🤖 TESTING RAG SYSTEM WITH GPT-2 GENERATOR")
print("="*80)

for question in test_questions:
    print(f"\n{'='*80}")
    print(f"❓ QUESTION: {question}")
    print("="*80)

    try:
        result = rag_answer(question, k=3, max_length=150)

        print(f"\n💡 ANSWER:\n{result['answer']}")

        print(f"\n📚 RETRIEVED DOCUMENTS:")
        for i, (doc, dist) in enumerate(zip(result['retrieved_docs'], result['distances'])):
            print(f"\n  Document {i+1} (distance: {dist:.4f}):")
            print(f"  {doc[:150]}...")

    except Exception as e:
        print(f"⚠️ Error: {str(e)}")

    print("\n" + "="*80)
````

````python
def ask_rag(question, show_sources=True):
    """
    Interactive function to ask questions to the RAG system

    Args:
        question: Your question
        show_sources: Whether to show retrieved documents
    """
    print(f"\n{'='*80}")
    print(f"Question: {question}")
    print("="*80)

    result = rag_answer(question, k=4, max_length=150)

    print(f"\nAnswer: {result['answer']}")

    if show_sources:
        print(f"\n{'─'*80}")
        print("Sources:")
        print("─"*80)
        for i, doc in enumerate(result['retrieved_docs']):
            print(f"\n[{i+1}] {doc[:200]}...")

    print("="*80)
    return result

# Try it!
ask_rag("What is deep learning?")
````

````python
# Evaluate RAG system performance
evaluation_questions = [
    {
        'question': "Who coined the term 'Artificial Intelligence'?",
        'expected_keyword': 'McCarthy'
    },
    {
        'question': "What is machine learning?",
        'expected_keyword': 'learn'
    },
    {
        'question': "When was AI founded?",
        'expected_keyword': '1956'
    },
    {
        'question': "What are neural networks?",
        'expected_keyword': 'network'
    }
]

print("📊 EVALUATING RAG SYSTEM")
print("="*80)

correct = 0
total = len(evaluation_questions)

for item in evaluation_questions:
    question = item['question']
    expected = item['expected_keyword'].lower()

    result = rag_answer(question, k=3, max_length=100)
    answer = result['answer'].lower()

    is_correct = expected in answer
    correct += int(is_correct)

    status = "✅" if is_correct else "❌"
    print(f"\n{status} Q: {question}")
    print(f"   Expected keyword: '{expected}'")
    print(f"   Found: {is_correct}")
    print(f"   Answer: {result['answer'][:100]}...")

print("\n" + "="*80)
print(f"📈 ACCURACY: {correct}/{total} ({correct/total*100:.1f}%)")
print("="*80)
````

````python
def pure_gpt_answer(question, max_length=150):
    """
    Generate answer using ONLY GPT-2 (no retrieval)
    """
    prompt = f"Question: {question}\n\nAnswer:"

    generated = generator(
        prompt,
        max_length=len(prompt.split()) + max_length,
        num_return_sequences=1,
        temperature=0.7,
        do_sample=True,
        pad_token_id=gpt_tokenizer.eos_token_id
    )

    answer = generated[0]['generated_text'].split("Answer:")[-1].strip()
    return answer

# Compare RAG vs Pure GPT-2
print("🔬 COMPARING RAG vs PURE GPT-2")
print("="*80)

test_question = "When was the term 'Artificial Intelligence' first coined?"

print(f"\nQuestion: {test_question}\n")

# RAG answer
rag_result = rag_answer(test_question, k=3, max_length=100)
print(f"📚 RAG Answer (with retrieval):")
print(f"{rag_result['answer']}\n")

# Pure GPT-2 answer
pure_answer = pure_gpt_answer(test_question, max_length=100)
print(f"🤖 Pure GPT-2 Answer (no retrieval):")
print(f"{pure_answer}\n")

print("="*80)
print("OBSERVATION:")
print("- RAG: Grounded in Wikipedia facts, accurate")
print("- Pure GPT-2: May hallucinate or provide incorrect information")
print("="*80)
````

# Exercise 4: Analyzing the Role of BERT in RAG Systems

## Chosen Application: Question Answering System

### How BERT is Used in the Retrieval Component

#### 1. Generating Document Embeddings
BERT processes each document chunk and generates dense vector representations:

**Process:**
```
Document: "Machine learning is a subset of AI that enables systems to learn from data."
    ↓
BERT Encoder (12 layers, multi-head attention)
    ↓
[CLS] Token Representation (768-dimensional vector)
    ↓
Document Embedding: [0.23, -0.45, 0.67, ..., 0.12]
```

**Key Features:**
- **Contextual**: Each word's embedding considers surrounding context
- **Bidirectional**: Reads text in both directions simultaneously
- **Semantic**: Captures meaning, not just keywords

#### 2. Generating Query Embeddings
User questions are embedded in the same vector space:

**Example:**
```
Query: "What is machine learning?"
    ↓
BERT Embedding: [0.24, -0.43, 0.65, ..., 0.11]
    ↓
Cosine Similarity with Document Embeddings
    ↓
Most Similar Documents Retrieved
```

#### 3. Vector Space Properties
- Documents with similar meanings have similar embeddings
- Query embeddings are comparable with document embeddings
- Distance in vector space represents semantic similarity

## Advantages of BERT Embeddings vs Other Methods

## Strengths and Weaknesses

### Traditional Generative Models (GPT)

#### Strengths ✅

1. **Fluency and Coherence**
   - Excellent natural language generation
   - Smooth, human-like responses
   - Strong grammatical correctness

2. **Reasoning and Creativity**
   - Can perform multi-step reasoning
   - Creative text generation
   - Good at explanation and elaboration

3. **Simplicity**
   - Single model, easier deployment
   - No retrieval infrastructure needed
   - Faster inference (no retrieval step)

4. **General Knowledge**
   - Broad knowledge across many domains
   - Good for common questions
   - Strong language understanding

5. **Zero-Shot Capabilities**
   - Can handle unseen tasks
   - Flexible instruction following
   - Adaptable to various prompts

#### Weaknesses ❌

1. **Factual Accuracy Issues**
   - **Hallucinations**: Makes up facts confidently
   - **Outdated Information**: Knowledge cutoff date
   - **No Source Attribution**: Can't cite sources

2. **Static Knowledge**
   - Can't access information after training
   - Requires retraining for updates
   - Expensive to keep current

3. **Domain-Specific Limitations**
   - Weaker on specialized domains
   - May lack depth in niche areas
   - Generic responses for specific questions

4. **Confidence Calibration**
   - Equally confident when wrong
   - No uncertainty indication
   - Difficult to verify claims

5. **Computational Cost**
   - Large models are expensive
   - High latency for complex queries
   - Significant memory requirements

### RAG Systems

#### Strengths ✅

1. **Factual Accuracy**
   - **Grounded in Sources**: Answers based on retrieved documents
   - **Reduced Hallucinations**: Context constrains generation
   - **Source Attribution**: Can show where information comes from

2. **Up-to-Date Information**
   - Access to current information
   - Update knowledge base without retraining
   - Can include recent documents

3. **Domain Expertise**
   - Excellent for specialized domains
   - Can use domain-specific knowledge bases
   - Detailed, accurate answers in niche areas

4. **Transparency**
   - Explainable: shows retrieved documents
   - Verifiable: users can check sources
   - Trustworthy: clear information provenance

5. **Scalability**
   - Add new documents easily
   - Knowledge grows independently of model
   - Cost-effective knowledge updates

#### Weaknesses ❌

1. **Retrieval Dependency**
   - Quality depends on retrieval accuracy
   - Bad retrieval → bad answers
   - May miss relevant information

2. **Complexity**
   - Two-stage system (more complex)
   - Requires vector database infrastructure
   - More components to maintain

3. **Latency**
   - Slower due to retrieval step
   - Two models to run (BERT + GPT)
   - Database query overhead

4. **Limited Reasoning**
   - Struggles with multi-hop reasoning
   - Less creative/flexible
   - Constrained by retrieved context

5. **Coverage Gaps**
   - Only as good as knowledge base
   - May not have relevant documents
   - Requires comprehensive data collection

# Exercise 6: Exploring the Future of RAG in NLP

## Recent Advancements in RAG (2024-2025)

### 1. Self-RAG: Self-Reflective Retrieval-Augmented Generation

**Innovation:**
- RAG system that can critique and improve its own outputs
- Generates special "reflection tokens" to evaluate retrieval quality
- Decides when to retrieve and when to use parametric knowledge

**Architecture:**
```
User Question
    ↓
Should I retrieve? [Self-reflection]
    ↓
    ├─ Yes → Retrieve documents → Generate with context
    │         ↓
    │    Is retrieval relevant? [Self-reflection]
    │         ↓
    │    Is answer supported? [Self-reflection]
    │
    └─ No → Generate directly from parameters
```

**Impact:**
- 10-15% accuracy improvement
- Reduces unnecessary retrieval
- Better handling of edge cases

**Source:** "Self-RAG: Learning to Retrieve, Generate, and Critique" (Asai et al., 2023)

### 2. REPLUG: Retrieval-Augmented Language Model Pre-Training

**Innovation:**
- Incorporates retrieval during pre-training (not just inference)
- Model learns to use retrieved context from the start
- Better integration of parametric and non-parametric knowledge

**Key Insight:**
```
Traditional RAG: Pre-train model → Add retrieval later
REPLUG: Pre-train WITH retrieval → Native retrieval capability
```

**Benefits:**
- +8-12% performance on knowledge-intensive tasks
- More natural integration of retrieved information
- Smaller models achieve better results

**Source:** "REPLUG: Retrieval-Augmented Language Model Pre-Training" (Meta AI, 2024)

### 3. Adaptive Retrieval (FLARE)

**Innovation:**
- Dynamic retrieval based on model confidence
- Only retrieves when uncertain
- Iterative refinement of answers

**Process:**
```
1. Start generating answer
2. Monitor confidence at each token
3. If confidence drops → trigger retrieval
4. Incorporate new information
5. Continue generation
```

**Advantages:**
- 40-60% fewer retrievals (cost savings)
- Maintains accuracy
- Better for conversational flow

**Source:** "Active Retrieval Augmented Generation" (Jiang et al., 2023)

### 4. Multi-Modal RAG

**Innovation:**
- Retrieval from text, images, videos, audio
- Cross-modal retrieval and generation
- Unified embedding space

**Example Application:**
```
Query: "Show me how to fix a leaky faucet"
    ↓
Retrieves:
- Text: Step-by-step instructions
- Images: Diagrams of faucet parts
- Video: Tutorial clips
    ↓
Generates:
- Multimodal response with integrated text and visuals
```

**Use Cases:**
- Medical diagnosis (text + medical images)
- E-commerce (product descriptions + images)
- Education (textbooks + diagrams + videos)

**Source:** "BLIP-2: Bootstrapping Language-Image Pre-training" (Li et al., 2023)
